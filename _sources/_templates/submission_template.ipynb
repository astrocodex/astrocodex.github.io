## Guidance for this template

This is a template you can use when submitting an exercise to the AstroCodEx. This demonstrates how you can set up your exercise using [MyST Markdown](https://mystmd.org/) for rendering on the website.

As part of each submission, we require
1.   A 1-3 sentence description of the exercise. This description should provide a high-level overview on what content and skills are covered.
2.   Learning objectives for the exercise. See [this resource](https://www.physport.org/recommendations/Entry.cfm?ID=125086) for guidance on writing learning objectives.
3. The intended audience (Early Undergraduate, Intermediate Undergraduate, Advanced Undergraduate, Graduate)
4.   A list of tags, selected from the list of available tags [here](here). If you would like your exercise to be listed under a new tag, see guidance [here](here).
5.   A list of required packages for the exercise, provided as a `requirements.txt` file. You may specify package versions.
6.   Solutions, which will be privately dissemenated by the AstroCodEx team by request.
7.   Any images or external data files relevant for the exercise.
8.   Any relevant instructor notes associated with the assignment.

Note that the header information in the cell above uses custom AstroCodEx website styling and may not render correctly when viewed locally on your machine. For example, the `learningobjective` class is defined specifically for the website environment.
